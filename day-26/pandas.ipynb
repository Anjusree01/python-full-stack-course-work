```python
import pandas as pd
```

```python
prices=[2999,10000,50000,4999]
products=["Wireless earbuds","Smartphone","laptop","smartwatch"]
product_prices=pd.Series(prices,index=products)
print(product_prices)
print("max=",product_prices.max())
print("min=",product_prices.min())
print("mean=",product_prices.mean())
print("sum=",product_prices.sum())
print("Head=\n",product_prices.head(2))
print("Tail=\n",product_prices.tail(2))
print("Sort values=\n",product_prices.sort_values())
print("Sort index=\n",product_prices.sort_index())#for descending order .sort_index(ascending=False)

print("map lambda=\n",product_prices.map(lambda x:f"₹{x}.00"))
print("map lambda=\n",product_prices.apply(lambda x:f"{x+(x*0.18)}"))
print("count values=\n",product_prices.value_counts())
```

```python
#dataframes
data={
"Product":["Wireless earbuds","Smartphone","Laptop","smartwatch","speaker"],
"Brand":["Soundmax","Technova","Bytecore","Timetravel","Echoboom"],
"Price":[2999,15999,52999,5000,6000],
"Stock":[50,30,20,40,60],
"Bestseller":[True,False,False,True,True]}
df=pd.DataFrame(data)
print(df)

print("Shape=\n",df.shape)
print("Columns=\n",df.columns)
print("Index=\n",df.index)

print("Dataframe info=\n",df.info())
print("Dataframe describe=\n",df.describe())

print("Head=\n",df.head(3))
print("Tail=\n",df.tail(2))

print("loc=\n",df.loc[0,"Product"])
print("iloc=\n",df.iloc[2,1])
print("iloc=\n",df.iloc[0])

print("---------usingloc---------\n",df.loc[df['Bestseller']==True])
print("---------usingloc---------\n",df.loc[df['Stock']<40])
print("---------usingloc---------\n",df.loc[df['Price']<10000])

#drop will not effect the original data
print("Dropped:\n",df.drop(columns=["Stock"]))
print(df)

print("rename=\n",df.rename(columns={"Product":"Product_name"},inplace=True))
print(df)



#groupby and agg
print("group_by=\n",df.groupby("Brand").agg({"Price":"sum","Stock":"sum"}))
print("group_by=\n",df.groupby("Brand").agg({"Price":["mean","max","min"]}))
print("group_by=\n",df.groupby("Brand")["Price"].mean())


```

```python
data2={
    "Brand":["Soundmax","Technova","Bytecore","Timetrack","Echoboom"],
    "Rating":[4.2,4.5,4.0,4.1,3.9],
    "Discount":[28,40,16,10,5]
}
df_ratings=pd.DataFrame(data2)
print(df_ratings)



#merge
df_merged=df.merge(df_ratings,on="Brand")
print(df_merged)
```
